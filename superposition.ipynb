```python
import torch
import torch.nn.functional as F
import math
from tqdm import tqdm

NUM_DIMS = 512
NUM_VECTORS = 1024
NUM_STEPS = 300

vectors = torch.randn(NUM_VECTORS, NUM_DIMS, requires_grad=True)
identity = torch.eye(NUM_VECTORS)

def cosine_similarity(vectors: torch.Tensor) -> torch.Tensor:
    x = F.normalize(vectors, p=2, dim=1)
    return x @ x.T

def loss(vectors: torch.Tensor) -> torch.Tensor:
    return torch.sum(torch.pow(cosine_similarity(vectors) - identity, 2))

optimizer = torch.optim.Adam([vectors], lr=0.05)

vector_history = [vectors.detach().clone().numpy()]

for i in tqdm(range(NUM_STEPS)):
    optimizer.zero_grad()
    loss(vectors).backward()
    optimizer.step()
    vector_history.append(vectors.detach().clone().numpy())

with torch.no_grad():
    similarity = cosine_similarity(vectors) - identity
    average_vector_similarity = torch.max(similarity, dim=0).values
    average_angle = torch.mean(torch.acos(average_vector_similarity) * 180 / math.pi)
    print("Average angle between closest vectors:", average_angle)
```
